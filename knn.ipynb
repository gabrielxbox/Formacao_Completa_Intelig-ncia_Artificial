```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import pandas as pd
import numpy as np
```

```python
mtcars = pd.read_csv('mt_cars.csv')
mtcars.head()
```

```python
X = mtcars[['mpg','hp']].values
y = mtcars['cyl'].values
```

```python
knn = KNeighborsClassifier(n_neighbors=3)
modelo =knn.fit(X,y)
```

```python
modelo
```

```python
y_prev = modelo.predict(X)
print(y_prev)
```

```python
accuracy = accuracy_score(y,y_prev)
precision = precision_score(y,y_prev,average='weighted')
recall = recall_score(y,y_prev,average='weighted')
f1 = f1_score(y,y_prev,average='weighted')
cm = confusion_matrix(y,y_prev)

print(f'Acuracia: {accuracy}, Precisão: {precision}, Recall: {recall}, F1:{f1}')
print('Matriz de Confusão:\n', cm)
```

```python
#'mpg', 'hp'
new_data = np.array([[19.3,105]])

predisao = modelo.predict(new_data)
predisao
#cyl

distanciaes, indices = modelo.kneighbors(new_data)
print(distanciaes)
print(indices)
```
