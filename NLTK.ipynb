```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer, SnowballStemmer, LancasterStemmer
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.tag import pos_tag, pos_tag_sents
import string
```

```python
nltk.download("stopwards")
nltk.download('stopwords')
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("tagsets_json")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("maxent_ne_chunker_tab")
nltk.download("wordnet")
nltk.download("words")
```

```python
texto = '''A vida é um breve instante entre dois silêncios.
Cada pensamento é uma centelha que ilumina o infinito.
Somos viajantes em um oceano de mistérios,
e cada pergunta nos aproxima do coração do cosmos.'''

print(texto)
```

```python
sentencas = sent_tokenize(texto)
print(sentencas)
```

```python
print(len(sentencas))
```

```python
token = word_tokenize(texto,language='portuguese')
print(token)
```

```python
print(len(token))
```

```python
stops = stopwords.words("portuguese")
print(stops)
```

```python
palavra_sem_stop = [p for p in token if p not in stops]
print(token)
print(palavra_sem_stop)
```

```python
print(string.punctuation)
```

```python
palavra_sem_pontuacao = [p for p in palavra_sem_stop if p not in string.punctuation]
print(palavra_sem_stop)
print(palavra_sem_pontuacao)
```

```python
frequencia = nltk.FreqDist(palavra_sem_pontuacao)
frequencia
```

```python
mais_comuns = frequencia.most_common(5)
mais_comuns
```

```python
stemmer = PorterStemmer()
stem1 = [stemmer.stem(word) for word in palavra_sem_pontuacao]
print(palavra_sem_pontuacao)
print(stem1)
```

```python
stemmer = SnowballStemmer("portuguese")
stem2 = [stemmer.stem(word) for word in palavra_sem_pontuacao]
print(palavra_sem_pontuacao)
print(stem1)
```

```python
spemmer3 = LancasterStemmer()
stem3 = [spemmer3.stem(word) for word in palavra_sem_pontuacao]
print(palavra_sem_pontuacao)
print(stem1)
```
