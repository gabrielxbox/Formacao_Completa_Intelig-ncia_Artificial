```python
import numpy as np
```

```python
import scipy.stats as stats
```

```python
data = np.array([30, 29, 30, 31, 100, 29, 28, 30, 300, 31, 32, 33])
```

```python
zscores = stats.zscore(data)
print(zscores)
```

```python
anomalia = data[np.abs(zscores) > 3]
print(anomalia)
```

```python
print(f"Anomlias Identificadas: {anomalia}")
```
