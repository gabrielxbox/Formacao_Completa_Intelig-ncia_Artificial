```python
import transformers
from transformers import pipeline
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Carregando modelo e tokenizador
model_name = "sshleifer/distilbart-cnn-12-6"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Texto para sumarizar
texto = """O modelo de linguagem natural é uma ferramenta poderosa na inteligência artificial moderna.
Ele permite que computadores entendam, gerem e manipulem a linguagem humana de maneira eficiente.
A tarefa de sumarização, especificamente, ajuda a reduzir textos longos em versões curtas, mantendo os pontos principais."""

# Tokenização e geração direta (bypass no pipeline)
inputs = tokenizer(texto, return_tensors="pt", truncation=True, max_length=1024)
summary_ids = model.generate(
    inputs["input_ids"],
    max_length=50,
    min_length=10,
    length_penalty=2.0,
    num_beams=4,
    early_stopping=True,
    forced_bos_token_id=None
)

# Decodificação do resultado
resultado = tokenizer.decode(summary_ids[0], skip_special_tokens=True)
print(resultado)
```
