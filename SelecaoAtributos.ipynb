```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
from sklearn.feature_selection import chi2, SelectKBest
```

```python
anucio = pd.read_csv('ad.data', header=None)
anucio.shape
```

```python
X = anucio.iloc[:, :-1].values
y = anucio.iloc[:, -1].values
```

```python
X_treinamento, Xteste, y_treinamento, yteste = train_test_split(X, y, test_size=0.3, random_state=0)
```

```python
modelo1 = GaussianNB()
modelo1.fit(X_treinamento, y_treinamento)
previsoes1 = modelo1.predict(Xteste)
print(accuracy_score(yteste, previsoes1))
```

```python
selecao = SelectKBest(chi2, k=7)
X_novo = selecao.fit_transform(X, y)
X_novo.shape
```

```python
print(selecao.get_support())
```

```python
X_treinamento, Xteste, y_treinamento, yteste = train_test_split(X_novo, y, test_size=0.3, random_state=0)
```

```python
modelo2 = GaussianNB()
modelo2.fit(X_treinamento, y_treinamento)
previsoes2 = modelo2.predict(Xteste)
print(accuracy_score(yteste, previsoes2))
```
