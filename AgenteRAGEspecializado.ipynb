```python
import os
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_community.vectorstores import FAISS
from langchain_community.document_loaders import PyPDFLoader
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain.chains import ConversationalRetrievalChain

```

```python
OPENAI_API_KEY = ""
os.environ["OPENAI_API_KEY"] = OPENAI_API_KEY
```

```python
loader = PyPDFLoader("manual_avancado_nobreak_fxp2000.pdf")
documents = loader.load()
```

```python
embaddings = OpenAIEmbeddings()
vectorstores = FAISS.from_documents(documents, embaddings)
retriever = vectorstores.as_retriever()
```

```python
chat = ChatOpenAI(model_name="gpt-4o", temperature=0)
```

```python
system_message = SystemMessage(content=('''
você é um assistemye virtual especializado do atendimento ao cliente no nobraak FP-2000.
responda claramente perguntas tecnicas, sobre as funcionalidades, garatia, manutenção, atualizações e suport tecnico.
se a pergunta for inrelevante, responda educadamente recusando a pergunta.
'''))
```

```python
store = {}
def get_session_history(session_id, str):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()
    return store[session_id]

    qa_chain = ConversationalRetrievalChain.from_llm(
    llm=chat,
    retriever=retriever,
    return_source_documents=False
    )

    chain_with_chat_history = RunnableWithMessageHistory(
        chain=qa_chain,
        history=get_session_history,
        input_messages_key="question",
        history_messages_key="chat_history"
    )
```

```python
session_id = "sessao1"
while true:
  pergunta = input("Você: ")
  if pergunta.lower() in ["sair", "quit","exit"]:
    print("Até logo!")
    break
  resposta = chain_with_chat_history.invoke(
      {"question": pergunta},
      config = {"configurable":{"session_id":session_id}}
      )
  print("Assistente> ", resposta["answer"])
```
