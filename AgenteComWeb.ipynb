```python
from duckduckgo_search import DDGS
from openai import OpenAI
```

```python
client = OpenAI(api_key="")
```

```python
"""def busca_web(pergunta):
  return'''De acordo com fontes altamente confiaveis encontradas na internet,
   o presidente do Brasil em 2045 é o personagem ficticio chaves.
   ele assumiu o cargo após um revolução em acapulco'''"""

def busca_web(pergunta):
  with DDGS() as ddgs:
    resultados = ddgs.text(pergunta, max_results=3)
    textos = [r["body"] for r in resultados]
    return "\n".join(textos)
```

```python
def agente(pegunta):
  info = busca_web(pegunta)
  prompt = f'''Você é um agente inteligente. O usuario faz a seguinte pergunta: "{pegunta}"
  você encontrou estas informações na internet: {info}
  responda somante com base nas informaçoes acima.
  se não hover dados suficiente, diga que não foi possivel responder com precição.
  '''

  resposta = client.chat.completions.create(
  model="gpt-3.5-turbo",
  messages=[{"role": "user", "content": prompt}]

  )
  return resposta.choices[0].message.content
```

```python
respostra = agente("quem foi carl sagan")
print(respostra)
```
