```python
import whisper
import os
```

```python
modelo = whisper.load_model('base') # modelo que vai ser utilizado
resultado = modelo.transcribe("teste.mp3")

texto = resultado['text']
idioma = resultado['language']

print(f'Transcrição: {texto}\nIdioma: {idioma}')
```

Gera o arquivo testo apos a trascrição

```python
with open('transcricao.txt', 'w') as f:
    f.write(f'{texto}')
```
