```python
from openai import OpenAI
from PIL import Image
import base64
import requests
from io import BytesIO
```

```python
client = OpenAI(api_key="")

response = client.chat.completions.create(
    model="dall-e-3",
    prompt = "centro de controle futurista com hologramas",
    size = "1024x1004",
    quality = "standard", #hd
    response_format = "b64_json",
    n=1
)
```

```python
image_url = response.data[0].url
print(image_url)
```

```python
image_response = request.get(image_url)
img = Image.open(BytesIO(image_response.cotent))
img.show()
```
